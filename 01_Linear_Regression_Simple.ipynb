```python
import numpy as np

x=np.array([1,2,3,4])
y=np.array([3,6,9,12])

m = (y[1] - y[0]) / (x[1] - x[0])
c = y[0] - (m * x[0])
m = round(m, 4)
c = round(c, 4)

if c == 0:
    equation = f"y = {m}x"
elif c > 0:
    equation = f"y = {m}x + {c}"
else:
    equation = f"y = {m}x - {abs(c)}"

print(equation)
```

```python
from sklearn.linear_model import LinearRegression
x=[[1],[2],[3],[4],[5]]
y=[[3],[6],[9],[12],[15]]
model=LinearRegression()
model.fit(x,y)
print(model.predict([[12]]))
```

```python
m=model.coef_[0]
c=model.intercept_
print('Slope: ',m)
print('Intercept: ',c)
print('Prediction: ', model.predict([[12]]))
```
